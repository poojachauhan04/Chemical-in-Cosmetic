# **Project Description** :-

This project is about exploring a dataset that lists **chemicals used in cosmetic products**, reported to the California Safe Cosmetics Program. The dataset has details about products, companies, brands, categories, and the chemicals inside them.

The main goal is to study how these chemicals are used, find patterns over time, and discover insights about product safety, company habits, and category behavior. By cleaning and organizing the data, the project aims to convert raw data into meaningful insights.

The first step is data preprocessing — fixing missing values, removing duplicates, and making sure dates are in the right format. **Feature engineering techniques** are applied to extract useful information such as reporting year and product status (active/discontinued).

The project then explores the data using **univariate, bivariate, and multivariate analysis techniques**. Various visualizations such as bar charts, histograms, boxplots, and line charts are used to understand distributions, relationships, and trends.

# **Key areas of analysis** :-
The project mainly looks at:

**Product categories** → How products are spread across different types and subtypes.

**Chemical usage** → Which chemicals are used most often.

**Company and brand activity** → Which companies and brands contribute the most.

**Trends over time** → How chemical usage changes year by year.

**Company practices** → Looking at which companies report more or fewer chemicals.



# **Objectives** :-

# **Chemicals in Cosmetics Dataset - Data Dictionary** :-

1-**CDPHId**

Unique identifier assigned by the California Department of Public Health (CDPH) for each record

2-**ProductName**

Name of the cosmetic product

3-**CSFId**

Unique ID for the reported product formulation (Cosmetic Safe Formulation ID)

4-**CSF**

Cosmetic formulation name or identifier submitted by the company

5-**CompanyId**

Unique ID assigned to each company

6-**CompanyName**

Name of the company that manufactures or distributes the product

7-**BrandName**

Brand under which the product is sold

8-**PrimaryCategoryId**

ID representing the main category of the cosmetic product

9-**PrimaryCategory**

Main category (e.g., Makeup, Skin Care, Hair Care)

10-**SubCategoryId**

ID representing the sub-category

11-**SubCategory**

Specific product type (e.g., Lipstick, Shampoo, Lotion)

12-**CasId**

Unique identifier for a chemical substance

13-**CasNumber**

CAS number – globally recognized chemical identifier

14-**ChemicalId**

Internal ID assigned to each chemical

15-**ChemicalName**

Name of the chemical present in the product

16-**InitialDateReported**

Date when the product-chemical combination was first reported

17-**MostRecentDateReported**

Latest date when the record was updated or confirmed

18-**DiscontinuedDate**

Date when the product was discontinued (if applicable)

19-**ChemicalCreatedAt**

Timestamp when the chemical record was created

20-**ChemicalUpdatedAt**

Timestamp when the chemical record was last updated

21-**ChemicalDateRemoved**

Date when the chemical was removed from the product (if applicable)

22-**ChemicalCount**

Number of chemicals reported for that product
